```python
# Import necessary libraries
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, f1_score, recall_score, classification_report
#Load dataset
cancer = datasets.load_breast_cancer()
```

```python
# print the names of the 13 features
print("Features: ", cancer.feature_names)

# print the label type of cancer('malignant' 'benign')
print("Labels: ", cancer.target_names)
```

```python
# Split dataset into training set and test set
X_train, X_test, y_train, y_test = train_test_split(cancer.data, cancer.target, test_size=0.3,random_state=109) # 70% training and 30% test
```

```python
# Create a Gradient Boosting Classifier
gb_clf = GradientBoostingClassifier(random_state=109)

# Train the model
gb_clf.fit(X_train, y_train)

# Make predictions on the test dataset
y_pred = gb_clf.predict(X_test)
```

```python
# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
```

```python
# Print evaluation metrics
print("Model Evaluation Metrics:")
print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
# Perform K-Fold Cross-Validation on the trained model
cv_scores = cross_val_score(gb_clf, X_train, y_train, cv=5, scoring='accuracy')
```

```python
# Print cross-validation results
print(f"\nCross-Validation Accuracy Scores: {cv_scores}")
print(f"Mean CV Accuracy: {cv_scores.mean():.4f}")
```

```python
# Define hyperparameter grid for tuning
param_grid = {
    'n_estimators': [50, 100, 200],  # Number of boosting stages
    'learning_rate': [0.01, 0.1, 0.2],  # Learning rate for tree updates
    'max_depth': [3, 4, 5]  # Maximum depth of each tree
}
```

```python
# Perform GridSearchCV for hyperparameter tuning
grid_search = GridSearchCV(GradientBoostingClassifier(random_state=109), param_grid, cv=5, scoring='accuracy')
grid_search.fit(X_train, y_train)

# Print the best parameters found by GridSearchCV
print("\nBest Parameters Found:")
print(grid_search.best_params_)
```

```python
# Train the model using the best parameters
best_gb_clf = grid_search.best_estimator_
best_gb_clf.fit(X_train, y_train)

# Make predictions using the tuned model
y_pred_tuned = best_gb_clf.predict(X_test)

# Evaluate the tuned model
accuracy_tuned = accuracy_score(y_test, y_pred_tuned)
precision_tuned = precision_score(y_test, y_pred_tuned)
recall_tuned = recall_score(y_test, y_pred_tuned)
f1_score_tuned = f1_score(y_test, y_pred_tuned)

# Print evaluation metrics for the tuned model
print("\nTuned Model Evaluation Metrics:")
print(f"Accuracy: {accuracy_tuned:.4f}")
print(f"Precision: {precision_tuned:.4f}")
print(f"Recall: {recall_tuned:.4f}")
print(f"F1-score: {f1_score_tuned:.4f}")  
print("\nClassification Report:\n", classification_report(y_test, y_pred_tuned))
```
